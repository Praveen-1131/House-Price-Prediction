# House Price Analysis
Explore the California Housing dataset, inspect data quality, visualize relationships, and understand the target before training models.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)
df = housing.frame
df.head()
```

## Dataset overview
The target column `MedHouseVal` is median house value in units of $100,000. The dataset is from California, not India.

```python
print('Shape:', df.shape)
display(df.describe().T)
print('Missing values per column:\n', df.isna().sum())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['MedHouseVal'], bins=40, kde=True)
plt.title('Distribution of Median House Value')
plt.xlabel('Median house value ($100,000 units)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm', center=0)
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.show()
```
